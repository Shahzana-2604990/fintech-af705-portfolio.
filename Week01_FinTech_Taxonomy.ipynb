## Classification Justification

Some FinTech companies provide several financial services, so their classification is based on their primary or most recognizable FinTech activity.

### 1. Plaid – RegTech

Plaid provides financial data connectivity and helps financial applications securely connect with users' bank accounts. I classified it under RegTech because its technology supports financial data access, compliance, and secure financial infrastructure.

### 2. Ant Group – Payments & Digital Wallets

Ant Group provides several financial technology services, but its best-known platform, Alipay, is strongly associated with digital payments and digital wallets. Therefore, I classified it under Payments & Digital Wallets.

### 3. Chainalysis – Blockchain & Crypto

Chainalysis provides blockchain analytics and cryptocurrency-related compliance services. Although it also supports compliance activities, its core technology is built around blockchain and cryptocurrency data, so I classified it under Blockchain & Crypto.

## FinTech Category Distribution

The following bar chart shows the number of companies classified into each FinTech taxonomy category.


```python
category_counts = df["Taxonomy Category"].value_counts()

plt.figure(figsize=(10, 6))

category_counts.plot(kind="bar")

plt.title("Number of FinTech Companies by Taxonomy Category")
plt.xlabel("FinTech Taxonomy Category")
plt.ylabel("Number of Companies")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

**Interpretation:** Payments & Digital Wallets is the most represented category in this dataset, which is broadly consistent with the importance of digital payments and wallet-based services in Pakistan's FinTech market.


## Reflection

The classification results provide useful evidence for understanding the FinTech landscape and connecting it with the lecture on Pakistan's financial technology market. In my dataset, Payments & Digital Wallets is the most represented category. This is broadly consistent with the discussion that digital payments are one of the most developed areas of FinTech in Pakistan. Pakistani platforms such as JazzCash, NayaPay and SadaPay are focused strongly on digital payments, while Easypaisa has expanded from digital financial services into digital retail banking. Pakistan's national Raast system also supports instant digital payments between individuals, businesses and government entities.

However, the results should not be interpreted as a complete measurement of the Pakistani FinTech industry because the dataset contains international companies and only a selected number of firms. The classification is also based mainly on each company's primary FinTech activity.

One comparatively under-represented category in the dataset is InsurTech. In Pakistan, one possible reason is the relatively smaller development of digital insurance compared with payments. Insurance products may require greater customer awareness, trust, regulatory processes and digital adoption. These factors can make the growth of InsurTech more gradual than payment services. Overall, the results support the lecture's discussion of the importance of payments while also showing that FinTech includes several other developing categories.
